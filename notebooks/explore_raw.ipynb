```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
f = ROOT / "data/raw/petrinex/Vol_2026-06-AB.CSV"

df = pd.read_csv(f, dtype=str, keep_default_na=False)
print(f.name, df.shape)

print(df["ActivityID"].replace("", "(blank)").value_counts())
print(df["FromToIDType"].replace("", "(blank)").value_counts())

w = df[df["FromToIDType"] == "WI"]
print(w.groupby("ActivityID").agg(
    rows=("FromToID", "size"),
    with_volume=("Volume", lambda s: (s.str.strip() != "").sum())))

print(w[w["Volume"].str.strip() != ""].head(2).T)

for c in ["Hours", "CCICode", "Volume"]:
    print(c, "仅空格:", (df[c].str.len().gt(0) & df[c].str.strip().eq("")).sum())
```

```python
p = ROOT / "data/raw/aer_st37/ST_37_Excel/ST37_ Production_Strings.xlsx"
x = pd.read_excel(p, dtype=str, nrows=5)
print(list(x.columns))
print(x.head())
```

```python
print(df["FromToIDType"].replace("", "(blank)").value_counts().to_string())

w = df[df["FromToIDType"] == "WI"]
t = w.groupby("ActivityID").agg(
    rows=("FromToID", "size"),
    with_volume=("Volume", lambda s: (s.str.strip() != "").sum()))
print(t.sort_values("rows", ascending=False).to_string())
```

```python
import re

def st37_to_petrinex(u):
    m = re.match(r"^(\w{2})/(\d{2})-(\d{2})-(\d{3})-(\d{2})W(\d)/(\d+)$", str(u).strip())
    if not m:
        return None
    le, lsd, sec, twp, rge, mer, seq = m.groups()
    return f"1{le}{lsd}{sec}{twp}{rge}W{mer}{int(seq):02d}"

ps = pd.read_excel(p, dtype=str)  # Read the entire file; this may take a while

ps["uwi_key"] = ps["Well_UWI"].map(st37_to_petrinex)

pet_wells = set(w["FromToIDIdentifier"].unique())

st_wells = set(ps["uwi_key"].dropna())

print("ST37 rows:", len(ps), " Conversion failures:", ps["uwi_key"].isna().sum())

print("Petrinex wells:", len(pet_wells), " Matched in ST37:", len(pet_wells & st_wells))

print("Match rate:", round(len(pet_wells & st_wells) / len(pet_wells), 3))

print("Examples of unmatched wells:", list(pet_wells - st_wells)[:5])
```

```python
# 1) Count how many rows have production for each ActivityID (top 10 only)
t = w.groupby("ActivityID").agg(
    rows=("FromToID", "size"),
    with_volume=("Volume", lambda s: (s.str.strip() != "").sum()))
print(t.sort_values("rows", ascending=False).head(10).to_string())

# 2) Recalculate the match rate after converting everything to uppercase
pet_wells_u = {x.upper() for x in pet_wells}
matched = pet_wells_u & st_wells
print("Match rate after uppercase conversion:", round(len(matched) / len(pet_wells_u), 4))

# 3) Distribution of unmatched wells by meridian
unmatched = pet_wells_u - st_wells
print("Number of unmatched wells:", len(unmatched))
print(pd.Series([s[-3] for s in unmatched]).value_counts().to_string())

print("Meridian distribution of all Petrinex wells:")
print(pd.Series([s[-3] for s in pet_wells_u]).value_counts().to_string())
```

```python
prod = df[(df["ActivityID"] == "PROD") & (df["FromToIDType"] == "WI")].copy()
prod["well_key"] = prod["FromToIDIdentifier"].str.upper()
prod["Volume_num"] = pd.to_numeric(prod["Volume"], errors="coerce")

prod["matched"] = prod["well_key"].isin(st_wells)
print("PROD rows:", len(prod), " Unmatched rows:", (~prod["matched"]).sum())

# Production share from unmatched wells by product
s = prod.groupby(["ProductID", "matched"])["Volume_num"].sum().unstack(fill_value=0)
s["unmatched_share"] = (s[False] / (s[False] + s[True])).round(4)
print(s.sort_values(False, ascending=False).head(6).to_string())

# Compare only the location portion (remove the leading 1 and the last two sequence digits)
# to check whether the difference is only in the sequence code
st_loc = {k[1:-2] for k in st_wells}
un = prod[~prod["matched"]]["well_key"].unique()
near = [k for k in un if k[1:-2] in st_loc]

print(
    "Unmatched wells:",
    len(un),
    " Location exists in ST37 (only differs by exception code/sequence):",
    len(near)
)

print(
    "Examples where the location cannot be found at all:",
    [k for k in un if k[1:-2] not in st_loc][:5]
)

# Number of rows per well in ST37 (used to determine the grain of dim_well)
print("ST37 rows:", len(ps), " Unique Well_UWI:", ps["Well_UWI"].nunique())
```

```python
bh = pd.read_excel(ROOT / "data/raw/aer_st37/ST_37_Excel/ST37_BH.xlsx", dtype=str)
print(len(bh), list(bh.columns)[:6])

bh["uwi_key"] = bh["Well_UWI"].map(st37_to_petrinex)
bh_keys = set(bh["uwi_key"].dropna())
print("BH conversion failures:", bh["uwi_key"].isna().sum())

un_set = set(un)   # Unmatched wells from the previous cell
print("Unmatched wells:", len(un_set), " Found in BH:", len(un_set & bh_keys))

# Operators with the most unmatched rows
u = prod[~prod["matched"]]
print(u.groupby("OperatorName").size().sort_values(ascending=False).head(5).to_string())
```

```python
print("Unique Well_UWI in BH:", bh["Well_UWI"].nunique(), " Total rows:", len(bh))

all_keys = st_wells | bh_keys
pet_prod_wells = set(prod["well_key"].unique())

print(
    "PROD wells:", len(pet_prod_wells),
    " Covered by union:", len(pet_prod_wells & all_keys),
    " Coverage rate:", round(len(pet_prod_wells & all_keys) / len(pet_prod_wells), 4)
)

# Share of production volume from remaining unmatched wells
prod["in_union"] = prod["well_key"].isin(all_keys)

r = prod.groupby(["ProductID", "in_union"])["Volume_num"].sum().unstack(fill_value=0)
r["missing_share"] = (r[False] / (r[False] + r[True])).round(4)

print(r.sort_values(False, ascending=False).head(6).to_string())

# Licence status of the 415 unmatched wells in BH
bh_un = bh[bh["uwi_key"].isin(un_set)]
print(bh_un["Licence_Status"].str.strip().value_counts().head(8).to_string())
```
